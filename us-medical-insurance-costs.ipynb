```python
import csv
data = []
with open ("insurance.csv", "r") as file:
    csv_reader = csv.DictReader(file)
    for row in csv_reader:
        data.append(row)
```

```python
men_data = []
women_data = []
for row in data:
    if row["sex"] == "male":
        men_data.append(row)
    elif row["sex"] == "female": 
        women_data.append(row)
def average(data, column_name):
    suma = 0.0
    counter = 0.0
    for row in data:
        suma += float(row[column_name]) 
        counter += 1
    return suma / counter
avg_mens_age = average(men_data, "age")
avg_womens_age = average(women_data, "age")
print(avg_mens_age, avg_womens_age)
```

```python
female_smoker = 0
male_smoker = 0
for row in data:
    sex = row["sex"]
    smoker = row["smoker"]
    if sex == "male":
        if smoker == "yes":
                male_smoker += 1
    elif sex == "female":
        if smoker == "yes":
                female_smoker += 1
print(female_smoker)
print(male_smoker)
```
